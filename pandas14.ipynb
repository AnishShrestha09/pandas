```python
# Given a positive integer N, repeatedly perform the following operation:
#- If .N.is.even, divide.it.by.2.
#-If .N.is.odd, subtract.1.from.it.
# Return.the number.of operations .required .to make.N.equal .to.zero.

# Input: 14
# .Output: 6
```

```python
Input = 14
count = 0
while Input != 0:
  # print(Input)
  if Input % 2 == 0:
    Input = Input/2

  else:
    Input = Input -1

  count = count + 1
print(count)
```

```python
# There are several chips placed at different positions on a number line.

# You can move a chip:
# - By 2 positions at a cost of 0.
# - By 1 position at a cost of 1.
# Find the minimum total cost required to move all chips to the same position.

# Input:
# position = [1, 2,3]
# Output: 0+1 = 1
```

```python
position = [1, 2,3,4,5,6]
lenA = len(position)
count = 0
for i in range(1,(lenA//2)+1):
  # print(i)
  for j in range(1,i+1):
    if i == j:
      count = count +1

print(count)
```

```python
Given a list of sentences,
find the maximum number of words present in any single sentence.


Words are separated by spaces.

"I love programming",
"Data structures are important",
"Practice daily"
```

```python
output = ["I love programming",
"Data structures are important",
"Practice daily"]

# outputtwo = output.split(" ")
# max = len(outputtwo[0])
max = 0

for i in output:
  outputtwo = i.split(" ")
  if len(outputtwo)> max:
    max = len(outputtwo)


print(max)
```
