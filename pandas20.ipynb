```python
# Problem: Merge two sorted arrays into the first array.
# The first array has enough empty space at the end.

# Input:
# nums1 = [1, 2,3,0,0,0]
# nums2 = [2,5,6]

# Output: [1, 2, 2, 3, 5, 6]
```

```python
nums1 = [1, 2,3,0,0,0]
nums2 = [2,5,6]
i = len(nums1)
j = len(nums2)
output = []

while ( i < len(nums1)):
  if j >= len(nums2) -1:
    break
  while(j < len(nums2)):
    if nums1[i] < nums2[j]:
      output.append(nums1[i])
      j += 1

    if nums1[i] > nums2[j]:
      output.append(nums2[j])
      i += 1
      break
print(output)
```
