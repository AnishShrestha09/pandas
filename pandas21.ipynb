```python
import pandas as pd

data = {
    "Product": [
        "Laptop", "Phone", "Tablet", "Mouse", "Keyboard",
        "Monitor", "Laptop", "Phone", "Mouse", "Monitor"
    ],
    "Category": [
        "Electronics", "Electronics", "Electronics", "Accessories", "Accessories",
        "Electronics", "Electronics", "Electronics", "Accessories", "Electronics"
    ],
    "Salesperson": [
        "Amit", "Neha", "Ravi", "Amit", "Neha",
        "Ravi", "Neha", "Amit", "Ravi", "Neha"
    ],
    "Revenue": [
        80000, 50000, 30000, 5000, 7000,
        45000, 90000, 60000, 8000, 45000
    ],
    "Units_Sold": [
        4, 5, 3, 10, 7,
        3, 5, 6, 12, 4
    ]
}

df = pd.DataFrame(data)

df
```

```python
# Which products generated the highest revenue?
df[df["Revenue"]==np.max(df["Revenue"])]["Product"]
```

```python
# Which products generated the lowest revenue?
df[df["Revenue"]==np.min(df["Revenue"])]["Product"]
```

```python
# Which products generated the highest revenue? show top 3

df.sort_values(by = "Revenue",ascending = False).head(3)
```

```python
# Which products generated the highest revenue? show top 5

df.sort_values(by = "Revenue",ascending = True).head(5)
```

```python
# Find the top 3 sales transactions
df.sort_values(by = "Units_Sold",ascending = False).head(3)

```

```python
# Sort by category and revenue -- > cat-asc, rev-desc


df.sort_values(by = ["Category","Revenue"],
               ascending = [True,False])
```
