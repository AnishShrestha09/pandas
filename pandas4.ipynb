```python
# Find the Frequency of Every Element

# Print each unique element along with its frequency.
# Input: 10 20 10 30 20 10
# output :
        # 10 3
#         20 2
#         30 1
```

```python
Input= [10 ,20, 10, 30, 20, 10]
output = {}

for i in Input:
  if i  not in output:
    output[i] = 1
  else:
    output[i]+=1


print(output)
```

```python
# Find the First Repeating Element

# Find the first element in the array that appears more than once.

# Input:
# 5 7 3 7 2 3

# Output: 7
```

```python
Input= [5, 7, 3, 7, 2, 3]
output= []

for i in Input:
  if i not in output:
    output.append(i)
  else:
    print(i)
    break
```

```python
# Find the First Non-Repeating Element

# Find the first element whose frequency is exactly 1.

# Input:
# 4 5 4 6 5 7 6

Output: 7
```

```python
Input=[ 4, 5, 4, 6, 5, 7, 6]
output = {}

for i in Input:
  if i  not in output:
    output[i] = 1
  else:
    output[i]+=1


for key in output:
  if(output[key]==1):
    print(key)
    break

```

```python
### Check Whether the Array is Sorted

Determine whether an array is sorted in ascending order.

**Input:**
`10 20 25 30 45`
```

```python
Input = [10, 20, 25, 30, 45,30]
flag = True
for i in range(0,len(Input)-1):
  if(Input[i]>Input[i+1]):
    flag = False


print(flag)
```
