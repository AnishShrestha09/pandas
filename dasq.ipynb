```python
# Find the Largest Difference Between Two Elements

# Find the maximum difference between two elements
# where the larger element occurs after the smaller element.

Input:
7 1 5 3 6 4

Output:
5
```

```python
import numpy as np
```

```python
arr = [7, 1, 5, 3 ,6, 4]

smallest = 10000
for i in arr[:-2]:
  for j in arr[1:]:
    diff = i - j
    if diff < smallest:
      smallest = diff

print(np.absolute(smallest))
```

```python
# Problem: Determine whether a number greater than 1 is prime.
# Use trial division only up to its square root.
```

```python
import numpy as np
```

```python
a = 70


prime = True
for i in range(2,int(np.sqrt(a))+1):
  if a%i == 0:
    print("not prime")
    prime = False
    break
  if(prime == True):
    print("prime")

```

```python
# Find the Minimum Absolute Sum of Two Elements

# Problem statement: Find two elements at different indices whose sum has the smallest absolute value.

# Example: [-8, 4, 5, -2, 10]
```

```python
Example=  [-8, 4, 5, -2, 10]

output = []


min = 1000

for i in range(len(Example)):
  if Example[i]<min:
    min = Example[i]


for i in range(len(Example)):
  if Example[i] != min:
    output.append(Example[i]+min)


abs = np.absolute(output)

np.min(abs)





```
