```python
# Find the index of the first element divisible by K

# Problem:
# Find the zero-based index of the first element divisible by K.
# If no such element exists, print -1. Assume K is nonzero.
```

```python
Example =  [7, 11, 18, 24, 30]
K = 6

```

```python
output = -1
for i in range(len(Example)):
  if Example[i]%K == 0:
    output = i
    break

print(output)
```

```python
# Find the Median of an array of integers.
import numpy as np
c = [4, 7,3,1,10,6]
a =np.sort(c)
b = len(c)
print(a)

if (b%2 !=0):
  middleindex = int(b/2)

  print(a[middleindex])
else:
  middleindex = int(b/2)
  print((a[middleindex]+ a[middleindex - 1])/2)
```
