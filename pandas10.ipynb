```python
import pandas as pd
import numpy as np

sales = pd.DataFrame( {
"Product": ["Laptop","Mouse", "Monitor", "Keyboard", "Printer"],
"Quantity": [5, 20, 4, 10, 3],
"Unit_Price": [50000, 800, 12000, 1500, 18000],
"Cost": [220000, 10000, 40000, 10000, 45000]
})
```

```python
sales['Revenue'] = sales['Quantity'] * sales['Unit_Price']
```

```python
sales
```

```python
#  find the profit

sales['profit'] = sales['Revenue'] - sales['Cost']
```

```python
sales["sales_cat"] = np.where(
   sales['profit']> 10000,
    "highProfit",
    "lowProfit"
)
```

```python
sales
```

```python
students = pd.DataFrame({
"Name": ["Amit", "Priya", "Rahul", "Sneha", "Karan"],
"Python": [85, 95, 45, 70, 55],
"SQL": [80, 90, 50, 65, 60],
"Statistics": [90, 92, 40, 75, 50]
})
```

```python
students
```

```python
# Problem: Calculate total marks.and percentage for three subjects.
# Assign grade .A for .a.percentage .of at least 80,
# B.for.at least.60.but.below.80, and .C.otherwise.
```

```python
students["total_marks"] = students[["Statistics","SQL","Python"]].sum(axis = 1)
```

```python
students
```

```python
# Assign grade A for a percentage of at least 80
# B.for.at least.60.but.below.80, and .C.otherwise.

students["percentage"] = np.round(students["total_marks"]/3,2)

students["Grade"] = np.where(
    students["percentage"]> 80,
    "A",
    np.where(
         (students["percentage"]>= 60) & (students["percentage"]<80),
    "B",
      "C"
    )
)

```

```python
students
```
