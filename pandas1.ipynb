```python
# Create a new list containing each value only once, preserving the order of its first appearance.

numbers = [4, 2, 4, 7, 2, 9, 7, 10]
howmany = []


for num in numbers:
  if num not in howmany:
      howmany.append(num)

print(howmany)
```

```python
# Given a list of integers, determine whether the list is sorted in ascending order.

# Do not use sort() or sorted().

# Input: [10, 20, 30, 40]
# Output: Sorted

# Input: [10, 30, 20, 40]
# Output: Not Sorted
```

```python
Input =  [10, 30, 20, 40]

num1 = Input[0]
flag = True

for input in Input[1:]:
  if input > num1 :
    num1 = input
  else:
    flag= False
    break

print(flag)
```
