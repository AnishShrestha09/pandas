```python
# find the smallest absolute difference between an element of the first array and an element of the second.

# Example: # Output: 1
A = [1, 4, 8, 10]
B = [2, 5,9,12]

smallest =float('inf')
for i in A:
  for j in B:
    diff = i - j
    if diff < 0:
      diff *= -1
    if smallest > diff:
      smallest = diff

print(smallest)

```
