```python
# Check whether two strings are anagrams

# Problem:
# Two strings are anagrams if they contain the same characters with the same frequencies, regardless of order.
# Assume the strings contain lowercase English letters.

# Example: Input: listen, silent|
# Output: True
```

```python
example1 = "aab"
example2 = "abb"

arr=[]
arr2 = []
anagram = True


for i in example1:
  arr.append(ord(i))

for j in example2:
  arr2.append(ord(j))



if len(arr)!=len(arr2):
  anagram= False
  print("not anagram")

else:
  for i in arr.copy():
    if i in arr2:
      arr.remove(i)
      arr2.remove(i)



  if arr == arr2:
    print("anagram")
  else:
    print("not anagram")
```
