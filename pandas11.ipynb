```python
# Replace every vowel with a star

# Problem statement:
# Replace each vowel (a, e, i, o, u, in either case) with *.
# Keep all other characters unchanged.

# Example: "Education"
# Output:
```

```python
Example =  "Education"
Output = ""

for i in Example.lower():
  if i in ["a","e","i","o","u"]:
    Output+="*"
  else:
    Output+=i
```

```python
Example = [4, 7,2,2]


unique = True
for i in range(0,len(Example)-1):
  for j in range(i+1,len(Example)):
    if Example[i] == Example[j]:
      unique = False
      break


print("uniquevalues"if unique else "noallunique")
```

```python
# Problem:
# Create an attendance status.
# \Students with attendance of at least 75% are Eligible; others have a Shortage.
# Filter eligible students and export the result to CSV.
import pandas as pd
import numpy as np

students = pd.DataFrame({
"Student_ID": [101, 102, 103, 104, 105, 106],
"Name": ["Amit", "Priya", "Rahul", "Sneha", "Karan", "Neha"],
"Attendance": [90, 72, 75, 60, 85, 74],
"Branch": ["CSE", "IT", "CSE", "ECE", "ME", "IT"]
})
```

```python
students["attendance_stat"] = np.where(
    students["Attendance"] >= 75,
    "Eligible",
    "Shortage"
)
```

```python
students
```

```python
elgiblestd = students[students["Attendance"]>= 75]
```

```python
elgiblestd.to_csv("eligble_student.csv",index = False)
```
