```python
# Problem: An array contains N distinct numbers from the range 0 to N,
# with one number missing. Find the missing number.

# Example: [3, 0,1]
# Output: 2
```

```python
Example = [4,2, 0,1]

for i in range(len(Example)+1):
  if i not in Example:
    print(i)
```

```python
Problem: Find the element that occurs more than n / 2 times.
Assume such an element is guaranteed to exist.



Output:
```

```python
Example = [3, 2, 3, 3, 1]

length = len(Example)//2

# for i in Example:
#   if i
#   \

output = []
for i in Example:
  count = Example.count(i)
  if count >=length:
    output.append(i)
print(max(output))


```

```python
Problem: Given a sorted array of distinct integers and a target,
return its index if found.
Otherwise, return the index where it should be inserted to maintain sorted order.

Example: arr = [1, 3, 5, 6], target = 2 - 1
```

```python
arr = [1,2, 3, 4,5, 7]
target = 6
output = []

if target < arr[0]:
  print(0)
else:
  for i in range(len(arr)):
    if arr[i] == target:
      output.append(i)
    elif target >arr[i] and target< arr[i+1]:
      output.append(i+1)

  if target not in arr:
    print(len(arr) )
  else:
    print(output)

```
