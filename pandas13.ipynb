```python
import pandas as pd
import numpy as np
data = {
"Student": ["Aman", "Riya", "Karan", "Neha", "Rohit"],
"Attendance": [85, 90, np.nan, 75, 80],
"Marks": [78, np.nan, 65, 88, np.nan],
"Study_Hours": [3, 4, 2, np.nan, 5]

}

df = pd.DataFrame(data)

df
```

```python
df.info()
```

```python
df.isna().sum()
```

```python
print("not null")
notnull = df.notna().sum().sum()
print(notnull)

print("total null")
null = df.isna().sum().sum()
print(null)
```

```python
# What % of values are missing in the entire dataframe

print((null/df.size)* 100)
```

```python
#what % of rows have some missing data?
total_missing = df.isna().any(axis = 1).sum()
```

```python
total_missing_perc =total_missing/ df.shape[0] * 100
print(total_missing_perc)
```
