```python
# Problem statement: Given asentence,
# reverse every individual .word.without .changing.the.order .of .the .words.

# Input: s="Hello.World.Python"
# Output: "olleH.dlroW.nohtyP"
```

```python
s="Hello World Python"
sentences = s.split()
reverse = []
for i in sentences:
  a = i[::-1]
  reverse.append(a)

print(" ".join(reverse))
```

```python
# Problem statement: Given a string, find.the length.of the longest substring
# that contains.no.repeated .characters.

# Input: s="abcabcbb"
# Output: 3
```

```python
s="pwwkew"
s2 = ""
max = 0

for i in s:
  if i not in s2:
    s2 += i
  else:
    print(s2)
    s2 = ""
    s2 = f'{i}'
  if max < len(s2):
    max = len(s2)

print(max)
```
