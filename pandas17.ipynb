```python
import pandas as pd

df = pd.DataFrame({
"Student": ["Amit", "Neha", "Ravi", "Priya", "Karan"],
"Age": [20, 19, 21, 20, 22],
"Marks": [85, 92, 76, 95, 88],
"Attendance": [90, 95, 80, 98, 85]
})
print(df)
```

```python
# Who has the highest marks?
import numpy as np

df[df["Marks"]==np.max(df["Marks"])]["Student"]
```

```python
# Who has the lowest attendance?
# - Ravi
```

```python
df[df["Marks"]==np.min(df["Marks"])]["Student"]
```

```python
# Who is the youngest student?
#- Neha
```

```python
df[df["Age"]==np.min(df["Age"])]["Student"]
```

```python
df.sort_values(by = "Marks",ascending=False)
```
